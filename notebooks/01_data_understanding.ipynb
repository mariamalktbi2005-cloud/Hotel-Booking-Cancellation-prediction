```python
#  INITIAL DATA QUALITY FINDINGS

# 1. Missing Values Analysis
missing_values = df.isnull().sum()

missing_percentage = (missing_values / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage
})

missing_table = missing_table[
    missing_table["Missing Values"] > 0
]

missing_table = missing_table.sort_values(
    by="Missing Values",
    ascending=False
)

print("Missing Values Analysis:")
display(missing_table)

# 2. Missing Values Visualization
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0]

missing_values.sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10, 5),
    title="Missing Values by Column"
)

plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Duplicate Records Analysis
duplicate_count = df.duplicated().sum()

duplicate_percentage = (duplicate_count / len(df)) * 100

print("\nDuplicate Records:", duplicate_count)
print("Duplicate Percentage:", round(duplicate_percentage, 2), "%")

# 4. Sample Duplicate Records
print("\nSample Duplicate Records:")
duplicates = df[df.duplicated()]

display(duplicates.head())
```

```python
#  DATA CLEANING

# 1. Before Cleaning
print("========== BEFORE CLEANING ==========")

print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())

# 2. Handling Missing Values

# Remove company because it has a very high percentage
# of missing values
df = df.drop(columns=["company"])

# Fill missing agent values with 0
df["agent"] = df["agent"].fillna(0)

# Fill missing country values with "Unknown"
df["country"] = df["country"].fillna("Unknown")

# Fill missing children values with 0
df["children"] = df["children"].fillna(0)

print("\nMissing values handled successfully!")

# 3. Removing Duplicate Records
df = df.drop_duplicates()

print("\nDuplicate Records After Cleaning:", df.duplicated().sum())

# 4. Converting Date Column
df["reservation_status_date"] = pd.to_datetime(
    df["reservation_status_date"],
    errors="coerce"
)

print("\nDate Column Type:")
print(df["reservation_status_date"].dtype)

# 5. After Cleaning
print("\n========== AFTER CLEANING ==========")

print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())

print("\nRemaining Missing Values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nFirst 5 Records of Cleaned Dataset:")
display(df.head())

# 6. Save Cleaned Dataset
df.to_csv("hotel_bookings_cleaned.csv", index=False)

print("\nCleaned dataset saved successfully!")
```

```python
#  INCONSISTENT VALUES

print("\n========== INCONSISTENT VALUES ==========")

# 1. Find "Undefined" labels
print("\n'Undefined' labels BEFORE fixing:")
for col in ["meal", "market_segment", "distribution_channel"]:
    print(col, ":", (df[col] == "Undefined").sum())

# 2. Find inconsistent country codes
# (3 letters is the normal ISO format, 2 letters is inconsistent)
country_length = df["country"].str.len()
print("\nCountry code lengths:")
print(country_length.value_counts())

print("\nCountry codes with 2 letters:")
print(df.loc[country_length == 2, "country"].value_counts())

# 3. Fix the inconsistent values

# "Undefined" meal means the same as "SC" (no meal package)
df["meal"] = df["meal"].replace("Undefined", "SC")

# Tiny number of rows, cannot be guessed, so mark as Unknown
df["market_segment"] = df["market_segment"].replace("Undefined", "Unknown")
df["distribution_channel"] = df["distribution_channel"].replace("Undefined", "Unknown")

# CN is China written with 2 letters, change to the 3-letter code
df["country"] = df["country"].replace("CN", "CHN")

# 4. Check after fixing
print("\n'Undefined' labels AFTER fixing:")
for col in ["meal", "market_segment", "distribution_channel"]:
    print(col, ":", (df[col] == "Undefined").sum())

print("\nCN codes AFTER fixing:", (df["country"] == "CN").sum())
```

```python
#  OUTLIERS

print("\n========== OUTLIERS ==========")

# 1. Detect outliers with the IQR rule
def iqr_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return (series < lower) | (series > upper)

outlier_columns = ["lead_time", "adr", "stays_in_week_nights",
                   "days_in_waiting_list", "booking_changes"]

outlier_table = pd.DataFrame({
    "Outliers": [iqr_outliers(df[c]).sum() for c in outlier_columns],
    "Percentage (%)": [round(iqr_outliers(df[c]).mean() * 100, 2) for c in outlier_columns]
}, index=outlier_columns)

print("\nOutliers found by the IQR rule:")
display(outlier_table)

# 2. Visualize outliers
fig, axes = plt.subplots(1, len(outlier_columns), figsize=(16, 4))
for ax, col in zip(axes, outlier_columns):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()

# 3. Find impossible values
total_guests = df["adults"] + df["children"] + df["babies"]

impossible = (
    (total_guests == 0)         # booking with no guests
    | (df["adr"] < 0)           # negative price
    | (df["adr"] > 1000)        # extreme price (5,400)
    | (df["adults"] > 10)       # more than 10 adults
    | (df["babies"] > 5)        # more than 5 babies
)

print("\nImpossible records breakdown:")
print("Zero guests      :", (total_guests == 0).sum())
print("Negative ADR     :", (df["adr"] < 0).sum())
print("ADR above 1000   :", (df["adr"] > 1000).sum())
print("More than 10 adults:", (df["adults"] > 10).sum())
print("More than 5 babies :", (df["babies"] > 5).sum())
print("Total impossible records:", impossible.sum())

# 4. Remove the impossible records
print("\nShape before removing:", df.shape)
df = df[~impossible].copy()
print("Shape after removing :", df.shape)

# 5. Keep unusual but possible rows, and mark them with flags
df["zero_night_flag"] = (
    (df["stays_in_weekend_nights"] + df["stays_in_week_nights"]) == 0
).astype(int)

df["adr_zero_flag"] = (df["adr"] == 0).astype(int)

print("\nZero-night bookings flagged:", df["zero_night_flag"].sum())
print("Zero-price (ADR = 0) bookings flagged:", df["adr_zero_flag"].sum())

# Convert children to a whole number
df["children"] = df["children"].astype(int)
```
