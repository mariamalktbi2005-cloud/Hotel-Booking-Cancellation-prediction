```python

clean_df = df.copy()

# Remove columns with too many missing values
clean_df.drop(columns=['company'], inplace=True)

# Handle missing values
clean_df['agent'] = clean_df['agent'].fillna(0)

clean_df['country'] = clean_df['country'].fillna('Unknown')

clean_df['children'] = clean_df['children'].fillna(0)

print("Missing values handled successfully!")
```

```python

# Fix meal values
clean_df['meal'] = clean_df['meal'].replace(
    'Undefined', 'SC'
)

# Fix market segment
clean_df['market_segment'] = clean_df['market_segment'].replace(
    'Undefined', 'Unknown'
)

# Fix distribution channel
clean_df['distribution_channel'] = clean_df['distribution_channel'].replace(
    'Undefined', 'Unknown'
)

# Fix country code
clean_df['country'] = clean_df['country'].replace(
    'CN', 'CHN'
)

print("Inconsistent values corrected!")
```

```python
# Identify impossible records

impossible_values = (
    ((clean_df['adults'] + clean_df['children'] +
      clean_df['babies']) == 0) |
    (clean_df['adr'] < 0) |
    (clean_df['adr'] > 1000) |
    (clean_df['adults'] > 10) |
    (clean_df['babies'] > 5)
)

print("Impossible records identified:",
      impossible_values.sum())
```

```python

before_outliers = len(clean_df)

impossible_values = (
    ((clean_df['adults'] + clean_df['children'] + clean_df['babies']) == 0) |
    (clean_df['adr'] < 0) |
    (clean_df['adr'] > 1000) |
    (clean_df['adults'] > 10) |
    (clean_df['babies'] > 5)
)

clean_df = clean_df[~impossible_values].copy()

print("Rows before removing impossible values:", before_outliers)
print("Rows after removing impossible values:", len(clean_df))
print("Impossible records removed:", before_outliers - len(clean_df))
```
